## Descriptive analysis functions

```python
def heatmap2d(arr: np.ndarray, cmap = 'viridis', max_values = 21):
    fig, ax = plt.subplots()
    cax = ax.imshow(arr, cmap = cmap, aspect = 'auto')
    ax.set_xticks(np.arange(max_values))
    ax.set_yticks(np.arange(max_values))

    ax.set_xticklabels(np.arange(max_values), rotation = 45, ha = "right")  # Rotate x-labels for better visibility
    ax.set_yticklabels(np.arange(max_values))

    cbar = fig.colorbar(cax)

    plt.show()
```
