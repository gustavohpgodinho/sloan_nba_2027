## functions competitive balance

```python
def get_match_probs(df, df_ref_time, df_measures, threshold):
    return df\
    .merge(df_ref_time, how = "right")\
    .assign(dif = lambda df: df["dif"].ffill().astype("int"))\
    .assign(hpts = lambda df: df["hpts"].ffill().astype("int"))\
    .assign(vpts = lambda df: df["vpts"].ffill().astype("int"))\
    .merge(df_measures, how = "left")\
    .assign(flag_unbalanced = lambda df: (df["prob_home"] >= threshold).astype("int") + (df["prob_away"] >= threshold).astype("int") + (df["prob_tie"] >= threshold).astype("int"))\
    .assign(cnt_unbalanced = lambda df: (df["flag_unbalanced"] == 1) * (df["flag_unbalanced"].shift(1) == 0))\
    .assign(cnt_unbalanced = lambda df: df["cnt_unbalanced"].astype("int"))\
    .assign(cnt_unbalanced = lambda df: df["cnt_unbalanced"].cumsum())

def get_match_probs2(df, df_ref_time, df_measures, threshold):
    return df\
    .merge(df_ref_time, how = "right")\
    .assign(dif = lambda df: df["dif"].ffill().astype("int"))\
    .assign(hpts = lambda df: df["hpts"].ffill().astype("int"))\
    .assign(vpts = lambda df: df["vpts"].ffill().astype("int"))\
    .merge(df_measures, how = "left")\
    .assign(flag_unbalanced_home = lambda df: (df["prob_home"] >= threshold).astype("int"))\
    .assign(cnt_unbalanced_home = lambda df: (df["flag_unbalanced_home"] == 1) * (df["flag_unbalanced_home"].shift(1) == 0))\
    .assign(cnt_unbalanced_home = lambda df: df["cnt_unbalanced_home"].astype("int"))\
    .assign(cnt_unbalanced_home = lambda df: df["cnt_unbalanced_home"].cumsum())\
    .assign(flag_unbalanced_away = lambda df: (df["prob_away"] >= threshold).astype("int"))\
    .assign(cnt_unbalanced_away = lambda df: (df["flag_unbalanced_away"] == 1) * (df["flag_unbalanced_away"].shift(1) == 0))\
    .assign(cnt_unbalanced_away = lambda df: df["cnt_unbalanced_away"].astype("int"))\
    .assign(cnt_unbalanced_away = lambda df: df["cnt_unbalanced_away"].cumsum())\
    .assign(flag_unbalanced_tie = lambda df: (df["prob_tie"] >= threshold).astype("int"))\
    .assign(cnt_unbalanced_tie = lambda df: (df["flag_unbalanced_tie"] == 1) * (df["flag_unbalanced_tie"].shift(1) == 0))\
    .assign(cnt_unbalanced_tie = lambda df: df["cnt_unbalanced_tie"].astype("int"))\
    .assign(cnt_unbalanced_tie = lambda df: df["cnt_unbalanced_tie"].cumsum())

def get_no_return_point(df, df_ref_time, df_measures, threshold):
    return get_match_probs(df = df, df_ref_time = df_ref_time, df_measures = df_measures, threshold = threshold)\
    .assign(max_cnt_unbalanced = lambda df: df["cnt_unbalanced"].max())\
    .query("cnt_unbalanced == max_cnt_unbalanced")\
    [["period", "clock", "rt_secs", "hpts", "vpts", "dif", "prob_home", "prob_away", "prob_tie"]]\
    .iloc[[0]]
```
